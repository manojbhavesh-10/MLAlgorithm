```python
## Adding new custom column by using functions
def categorize_athlete(row):
    if row['height_cm'] < 175 and row['weight_kg'] < 70:
        return 'Lightweight'
    elif row['height_cm'] < 185 or row['weight_kg'] <= 80:
        return 'Middleweight'
    else:
        return 'Heavyweight'

bios['Category'] = bios.apply(categorize_athlete, axis=1)
```

### Advanced Functionality
##### .shift() .rank() .rolling() cumsum().

```python
## shift() -> Get the data of previous 1st rows and stored it in 2nd row Tuesday [yesterday_revenue] 
coffee['yesterday_revenue'] = coffee['revenue'].shift(2)
```

```python
coffee
```

```python
coffee['pct_change'] = coffee['revenue'] / coffee['yesterday_revenue'] * 100
```

```python
coffee
```

```python
bios.head()
```

```python
bios['height_rank'] = bios['height_cm'].rank()
```

```python
bios.sort_values(['height_rank']).sample(10)[['name', 'height_rank']]
```

```python
coffee.info
```

```python
## cumsum() -> return previous_data + current_data, add it in the current row
## for example index: 1 -> 99.75 + 89.85 = 189.6
coffee['cummulative_revenue'] = coffee['revenue'].cumsum()
```

```python
coffee
```

```python
## rolling() -> cummulative sum of three days
## index: 5-> 15.000 + 28.333 + 25.000 = 68.3333
latte = coffee[coffee['Coffee Type'] == 'Latte'].copy()
latte['3day'] = latte['Units Sold'].rolling(3).sum()
```

```python
latte
```
