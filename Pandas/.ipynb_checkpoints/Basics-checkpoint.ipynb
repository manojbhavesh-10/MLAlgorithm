```python
import pandas as pd
```

```python
# create a new dataframe and reasign a columns as ['A', 'B', 'C']
df = pd.DataFrame([[1, 2, 3], [4, 5, 6], [7, 8, 9]], columns = ['A', 'B', 'C'])
```

```python
## head() function return the first 5 data and tail() is also there. return last 5 item
df.head()
```

```python
## index represent the row by default it starts from 0, 1, 2 and goes on.
df.index
```

```python
## converting to list to view our index
df.index.tolist()
```

```python
## We can reassign the index values.
df = pd.DataFrame([[1, 2, 3], [4, 5, 6], [7, 8, 9]], columns = ['A', 'B', 'C'], index=['x', 'y', 'z'])
```

```python
df
```

```python
## return the information about the data
df.info()
```

```python
## describe functions return the usefull data like (count, mean, std, min, max etc...)

df.describe()
```

```python
## return the shape of the dataframe
df.shape
```

```python
## nunique -> number unique, it returns the total number of unique values in each columns
df.nunique()
```

```python
## unique() -> returns the list of unique items present in the specific columns
df['A'].unique()
```
